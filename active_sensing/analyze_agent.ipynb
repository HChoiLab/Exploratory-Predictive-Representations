# Straightness

```python
from utils.plotting import *
def compute_monotonic_trajectory_tortuosity(
    mouse_sources: dict,
    mouse: str,
    z_key: str = "z_p",
    group_frac=(0.0, 1.0),
    exclude_first_n: int = 0,
    direction: str = "both",   # "in", "out", or "both"
    min_run_nodes: int = 2,     # minimum number of nodes in a run
    eps: float = 1e-8,
):
    """
    Compute tortuosity score on straight monotonic trajectories only.

    For each maximal monotonic run in node index:
        path_length / displacement
        = sum ||z_t - z_{t-1}|| / ||z_T - z_0||

    Monotonic run:
    - strictly increasing node index  -> outward
    - strictly decreasing node index  -> inward
    - breaks on reversal

    Parameters
    ----------
    direction : str
        "out"  -> only increasing runs
        "in"   -> only decreasing runs
        "both" -> use both

    min_run_nodes : int
        Minimum number of nodes in a run.
        2 means at least one transition.
        3 means at least two transitions, etc.

    Returns
    -------
    float
        Mean tortuosity across valid monotonic runs.
    """
    import numpy as np
    from pathlib import Path

    if mouse not in mouse_sources:
        raise KeyError(f"mouse='{mouse}' not in mouse_sources")

    if direction not in {"in", "out", "both"}:
        raise ValueError("direction must be 'in', 'out', or 'both'")

    if min_run_nodes < 2:
        raise ValueError("min_run_nodes must be >= 2")

    def extract_monotonic_runs(nodes):
        """
        Return list of (start_idx, end_idx, run_dir), inclusive endpoints.

        run_dir is:
        - "out" for strictly increasing
        - "in"  for strictly decreasing
        """
        nodes = np.asarray(nodes).astype(int)
        T = len(nodes)
        runs = []

        i = 0
        while i < T - 1:
            diff = nodes[i + 1] - nodes[i]

            if diff > 0:
                run_dir = "out"
            elif diff < 0:
                run_dir = "in"
            else:
                i += 1
                continue

            j = i + 1
            while j < T - 1:
                next_diff = nodes[j + 1] - nodes[j]
                if run_dir == "out" and next_diff > 0:
                    j += 1
                elif run_dir == "in" and next_diff < 0:
                    j += 1
                else:
                    break

            if (j - i + 1) >= min_run_nodes:
                runs.append((i, j, run_dir))

            i = j

        return runs

    metrics_dir = Path(mouse_sources[mouse]["metrics_dir"])
    bouts_all = load_bouts_from_dir(metrics_dir)
    bouts_sel, (start, end, N) = select_bouts_by_percent(bouts_all, group_frac)

    if len(bouts_sel) == 0:
        raise ValueError("No bouts selected.")

    ratios = []

    for b in bouts_sel:
        if "target_node" not in b or z_key not in b:
            continue

        nodes = np.asarray(b["target_node"]).astype(int)
        Z = np.asarray(b[z_key])

        if len(nodes) != Z.shape[0]:
            raise ValueError(
                f"Length mismatch: len(target_node)={len(nodes)} vs {z_key}.shape[0]={Z.shape[0]}"
            )

        if len(nodes) <= exclude_first_n:
            continue

        nodes = nodes[exclude_first_n:]
        Z = Z[exclude_first_n:]

        if len(nodes) < 2:
            continue

        runs = extract_monotonic_runs(nodes)

        for s, e, run_dir in runs:
            if direction != "both" and run_dir != direction:
                continue

            Z_run = Z[s:e + 1]

            if len(Z_run) < 2:
                continue

            path_length = np.sum(np.linalg.norm(Z_run[1:] - Z_run[:-1], axis=1))
            displacement = np.linalg.norm(Z_run[-1] - Z_run[0])

            if displacement < eps:
                continue

            ratios.append(path_length / displacement)

    if len(ratios) == 0:
        raise ValueError("No valid monotonic trajectories found.")

    return float(np.mean(ratios))
```

# Depth gradient 

```python
def compute_monotonic_depth_spearman(
    mouse_sources: dict,
    mouse: str,
    z_key: str = "z_p",
    group_frac=(0.0, 1.0),
    exclude_first_n: int = 0,
    direction: str = "both",   # "in", "out", or "both"
    min_run_nodes: int = 3,    # at least 3 nodes so Spearman is meaningful
    average_mode: str = "mean",  # "mean" or "median" across runs
    abs_value: bool = True,      # use abs(rho) by default
    eps: float = 1e-8,
):
    """
    Compute depth-monotonicity score on straight monotonic trajectories only.

    For each maximal monotonic run in node sequence:
      1) take latent vectors Z_run
      2) project them onto the first principal component (PC1)
      3) compute Spearman correlation between PC1 coordinate and maze depth

    This measures whether the latent trajectory is ordered along the
    maze hierarchy, separately for inward / outward / both.

    IMPORTANT:
    - We do NOT average z_p per node.
    - z_p is transition-context-like, so multiple z_p for the same node
      may be meaningfully different.

    Parameters
    ----------
    direction : str
        "out"  -> only increasing node-index runs
        "in"   -> only decreasing node-index runs
        "both" -> use both

    min_run_nodes : int
        Minimum number of nodes in a run.
        Use >= 3 for stable Spearman.

    average_mode : str
        "mean" or "median" across runs

    abs_value : bool
        If True, returns mean |rho| across runs.
        If False, returns signed rho after orienting PC1 so outward runs
        tend to be positive with increasing depth.

    Returns
    -------
    float
        Mean or median Spearman rho across valid monotonic runs.
    """
    import numpy as np
    from pathlib import Path

    try:
        from scipy.stats import spearmanr
    except ImportError:
        raise ImportError("scipy is required for spearmanr")

    if mouse not in mouse_sources:
        raise KeyError(f"mouse='{mouse}' not in mouse_sources")

    if direction not in {"in", "out", "both"}:
        raise ValueError("direction must be 'in', 'out', or 'both'")

    if min_run_nodes < 3:
        raise ValueError("min_run_nodes must be >= 3 for Spearman correlation")

    if average_mode not in {"mean", "median"}:
        raise ValueError("average_mode must be 'mean' or 'median'")

    def node_depth(node_idx: np.ndarray) -> np.ndarray:
        """
        Depth in binary tree:
        root=0, children=1, etc.
        For node n, depth = floor(log2(n+1))
        """
        node_idx = np.asarray(node_idx).astype(np.int64)
        return np.floor(np.log2(node_idx + 1)).astype(np.int64)

    def extract_monotonic_runs(nodes):
        """
        Return list of (start_idx, end_idx, run_dir), inclusive endpoints.

        run_dir:
        - "out" for strictly increasing node index
        - "in"  for strictly decreasing node index
        """
        nodes = np.asarray(nodes).astype(int)
        T = len(nodes)
        runs = []

        i = 0
        while i < T - 1:
            diff = nodes[i + 1] - nodes[i]

            if diff > 0:
                run_dir = "out"
            elif diff < 0:
                run_dir = "in"
            else:
                i += 1
                continue

            j = i + 1
            while j < T - 1:
                next_diff = nodes[j + 1] - nodes[j]
                if run_dir == "out" and next_diff > 0:
                    j += 1
                elif run_dir == "in" and next_diff < 0:
                    j += 1
                else:
                    break

            if (j - i + 1) >= min_run_nodes:
                runs.append((i, j, run_dir))

            i = j

        return runs

    def pc1_projection(Z_run: np.ndarray) -> np.ndarray:
        """
        Project latent points onto first principal component.
        Returns shape (L,)
        """
        Zc = Z_run - Z_run.mean(axis=0, keepdims=True)

        # SVD-based PCA
        _, _, Vt = np.linalg.svd(Zc, full_matrices=False)
        pc1 = Vt[0]  # (z_dim,)

        coord = Zc @ pc1  # (L,)
        return coord

    metrics_dir = Path(mouse_sources[mouse]["metrics_dir"])
    bouts_all = load_bouts_from_dir(metrics_dir)
    bouts_sel, (start, end, N) = select_bouts_by_percent(bouts_all, group_frac)

    if len(bouts_sel) == 0:
        raise ValueError("No bouts selected.")

    rho_list = []

    for b in bouts_sel:
        if "target_node" not in b or z_key not in b:
            continue

        nodes = np.asarray(b["target_node"]).astype(int)
        Z = np.asarray(b[z_key])

        if len(nodes) != Z.shape[0]:
            raise ValueError(
                f"Length mismatch: len(target_node)={len(nodes)} vs {z_key}.shape[0]={Z.shape[0]}"
            )

        if len(nodes) <= exclude_first_n:
            continue

        nodes = nodes[exclude_first_n:]
        Z = Z[exclude_first_n:]

        if len(nodes) < min_run_nodes:
            continue

        runs = extract_monotonic_runs(nodes)

        for s, e, run_dir in runs:
            if direction != "both" and run_dir != direction:
                continue

            nodes_run = nodes[s:e + 1]
            Z_run = Z[s:e + 1]

            if len(nodes_run) < min_run_nodes:
                continue

            depths = node_depth(nodes_run)

            # skip degenerate runs with no depth variation
            if np.all(depths == depths[0]):
                continue

            coord = pc1_projection(Z_run)

            # skip degenerate latent projection
            if np.std(coord) < eps:
                continue

            rho, _ = spearmanr(coord, depths)

            if np.isnan(rho):
                continue

            # orient sign so outward runs are positive when depth increases
            if not abs_value:
                if run_dir == "out" and rho < 0:
                    rho = -rho
                elif run_dir == "in" and rho > 0:
                    rho = -rho
            else:
                rho = abs(rho)

            rho_list.append(float(rho))

    if len(rho_list) == 0:
        raise ValueError("No valid monotonic trajectories found for depth-Spearman metric.")

    if average_mode == "mean":
        return float(np.mean(rho_list))
    else:
        return float(np.median(rho_list))
```

# Separation with hyperplane


```python
def compute_directional_branch_separability(
    mouse_sources: dict,
    mouse: str,
    z_key: str = "z_p",
    group_frac=(0.0, 1.0),
    exclude_first_n: int = 0,
    n_splits: int = 5,
    balance_classes: bool = True,
    random_state: int = 0,
):
    """
    Compute directional branch separability of latent transition points.

    Each valid transition contributes one point:
        point = z_t
        label = 1 if node_t < node_{t+1}  (outward)
                0 if node_t > node_{t+1}  (inward)

    Then train a linear classifier to separate inward vs outward points.

    Returns
    -------
    float
        Cross-validated balanced accuracy.

    Interpretation
    --------------
    ~0.5 : inward/outward branches are not linearly separable
    ~1.0 : inward/outward branches are cleanly separable
    """
    import numpy as np
    from pathlib import Path
    from sklearn.linear_model import LogisticRegression
    from sklearn.model_selection import StratifiedKFold
    from sklearn.metrics import balanced_accuracy_score
    from sklearn.preprocessing import StandardScaler
    from sklearn.pipeline import make_pipeline

    if mouse not in mouse_sources:
        raise KeyError(f"mouse='{mouse}' not in mouse_sources")

    if n_splits < 2:
        raise ValueError("n_splits must be >= 2")

    metrics_dir = Path(mouse_sources[mouse]["metrics_dir"])
    bouts_all = load_bouts_from_dir(metrics_dir)
    bouts_sel, _ = select_bouts_by_percent(bouts_all, group_frac)

    if len(bouts_sel) == 0:
        raise ValueError("No bouts selected.")

    X = []
    y = []

    for b in bouts_sel:
        if "target_node" not in b or z_key not in b:
            continue

        nodes = np.asarray(b["target_node"]).astype(int)
        Z = np.asarray(b[z_key])

        if len(nodes) != Z.shape[0]:
            raise ValueError(
                f"Length mismatch: len(target_node)={len(nodes)} "
                f"vs {z_key}.shape[0]={Z.shape[0]}"
            )

        if len(nodes) <= exclude_first_n + 1:
            continue

        nodes = nodes[exclude_first_n:]
        Z = Z[exclude_first_n:]

        if len(nodes) < 2:
            continue

        leaf_min = 31
        leaf_max = 126

        for t in range(len(nodes) - 1):
            x0 = int(nodes[t])
            x1 = int(nodes[t + 1])

            # keep only transitions involving terminal / leaf nodes 63-126
            if not ((leaf_min <= x0 <= leaf_max) or (leaf_min <= x1 <= leaf_max)):
                continue

            if x0 < x1:
                label = 1   # outward
            elif x0 > x1:
                label = 0   # inward
            else:
                continue

            X.append(Z[t])
            y.append(label)

    if len(X) == 0:
        raise ValueError("No valid inward/outward transitions found.")

    X = np.asarray(X, dtype=float)
    y = np.asarray(y, dtype=int)

    n_out = np.sum(y == 1)
    n_in = np.sum(y == 0)

    if n_out == 0 or n_in == 0:
        raise ValueError("Need both inward and outward transitions.")

    # Optional class balancing before CV.
    # Balanced accuracy already handles imbalance, but this avoids huge density bias.
    if balance_classes:
        rng = np.random.default_rng(random_state)
        idx_out = np.where(y == 1)[0]
        idx_in = np.where(y == 0)[0]
        m = min(len(idx_out), len(idx_in))

        keep = np.concatenate([
            rng.choice(idx_out, size=m, replace=False),
            rng.choice(idx_in, size=m, replace=False),
        ])
        rng.shuffle(keep)

        X = X[keep]
        y = y[keep]

    min_class_count = min(np.sum(y == 0), np.sum(y == 1))
    if min_class_count < 2:
        raise ValueError("Need at least 2 samples per class for cross-validation.")

    n_splits_eff = min(n_splits, int(min_class_count))

    clf = make_pipeline(
        StandardScaler(),
        LogisticRegression(
            penalty="l2",
            C=1.0,
            solver="liblinear",
            class_weight="balanced",
            random_state=random_state,
            max_iter=1000,
        ),
    )

    cv = StratifiedKFold(
        n_splits=n_splits_eff,
        shuffle=True,
        random_state=random_state,
    )

    scores = []

    for train_idx, test_idx in cv.split(X, y):
        clf.fit(X[train_idx], y[train_idx])
        y_pred = clf.predict(X[test_idx])
        score = balanced_accuracy_score(y[test_idx], y_pred)
        scores.append(score)

    return float(np.mean(scores))
```

```python
import pickle
import numpy as np
import matplotlib.pyplot as plt

with open("figs/res_seeds_separation_hyperplane.pkl", "rb") as f:
    dict1 = pickle.load(f)

with open("figs/agents_res_seeds_trans_consistency.pkl", "rb") as f:
    dict2 = pickle.load(f)

rew_p_lst = ["0.0", "0.01", "0.05", "0.1", "0.15", "0.3", "0.5", "0.85"]

# Combine per seed and mouse: combined[seed][mouse] = dict1[seed][mouse] * dict2[seed][mouse]
combined = {
    seed: {
        mouse: np.array(dict1[seed][mouse]) * np.array(dict2[seed][mouse])
        for mouse in dict1[seed]
        if seed in dict2 and mouse in dict2[seed]
    }
    for seed in dict1
    if seed in dict2
}

# Average across seeds for each mouse
mouse_mean = {
    mouse: np.mean(
        [combined[seed][mouse] for seed in combined if mouse in combined[seed]],
        axis=0
    )
    for mouse in next(iter(combined.values()))
}

mouse_std = {
    mouse: np.std(
        [combined[seed][mouse] for seed in combined if mouse in combined[seed]],
        axis=0
    )
    for mouse in next(iter(combined.values()))
}

mean_last = np.array([mouse_mean[f"Agent_{rew_p}"][-1] for rew_p in rew_p_lst])
std_last = np.array([mouse_std[f"Agent_{rew_p}"][-1] for rew_p in rew_p_lst])

x = np.arange(len(rew_p_lst))

plt.plot(x, mean_last, "-o", label="Mean")
plt.fill_between(x, mean_last - std_last, mean_last + std_last, alpha=0.3, label="Std")

plt.xticks(x, rew_p_lst)
plt.xlabel("Reward Probability")
plt.ylabel("Combined score")
plt.legend()
plt.tight_layout()
plt.show()
```
